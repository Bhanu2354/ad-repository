```python
##one hot encoding
data=pd.get_dummies(data,columns=['Diabetic'],drop_first=True)
data.head()
```

```python
##convert binary categorical variables into numerical values
binary_columns=["Smoking","AlcoholDrinking","Stroke","DiffWalking",'PhysicalActivity',"Asthma",'KidneyDisease', 'SkinCancer']
for col in binary_columns:
    data[col]=data[col].map({'Yes':1,"No":0})
```

```python
data.head()
```

```python
data=pd.get_dummies(data,columns=['Sex',"AgeCategory","Asthma","GenHealth"],drop_first=True)
data.head()
```

```python
##null values
data.isnull().sum()
```

```python
## define X input and Y target 
X=data.drop(columns=['HeartDisease'])
y=data['HeartDisease'].map({"Yes":1,"No":0})
```

```python
##train-test split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
clf=DecisionTreeClassifier(max_depth=5,criterion='gini',random_state=42)
clf.fit(X_train,y_train)
```
